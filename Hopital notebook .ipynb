```python
# Create a copy
df_clean = df.copy()
```

```python
# Check duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
# Split the Name column into Name and Patient ID
df[["Name", "Patient_ID"]] = df["Name"].str.rsplit(" ", n=1, expand=True)

# Convert Patient_ID to numeric
df["Patient_ID"] = pd.to_numeric(df["Patient_ID"], errors="coerce")

# Display the result
display(df.head())
```

```python
print(df.columns.tolist())
display(df[["Name", "Patient_ID"]].head())
```

```python
print("duplicate before:",df.duplicated().sum())
df = df.drop_duplicates()
print("Duplicates after:", df.duplicated().sum())
```

```python
 print(df.isnull().sum())
```

```python
print("Rows and columns:", df.shape)
display(df.head())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df.info()
```

```python
invalid_dates = df[
    df["Discharge Date"] < df["Date of Admission"]
]

display(invalid_dates)
print("Invalid date records:", len(invalid_dates))
```

```python
negative_bills = df[df["Billing Amount"] < 0]

display(negative_bills)
print("Negative billing records:", len(negative_bills))
```

```python
df = df[df["Discharge Date"] >= df["Date of Admission"]].copy()

# Reset the index
df = df.reset_index(drop=True)

print("Dataset after removing invalid dates:", df.shape)

display(df.head())
```

```python
df = df[df["Billing Amount"] >= 0].copy()
df = df.reset_index(drop=True)
```

```python
negative_bills = df[df["Billing Amount"] < 0]

display(negative_bills)
print("Negative billing records:", len(negative_bills))
```

```python
print("Dataset shape:", df.shape)
```

```python
df.to_excel(
    r"C:\Users\HP\Desktop\hospital1_cleaned.xlsx",
    index=False
)

print("Cleaned dataset saved successfully!")
```

```python
import seaborn as sns
```

```python

df = pd.read_excel(r"C:\Users\HP\Desktop\hospital1.xlsx")
```

```python
sns.set_theme(style="whitegrid")
```

```python
# Graph 1: Medical Condition
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="Medical Condition",
            y="Billing Amount", hue="Medical Condition",
            legend=False)
plt.title("Average Billing by Medical Condition")
plt.xticks(rotation=30)
plt.show()
```

```python
df["Month"] = pd.to_datetime(
    df["Date of Admission"]
).dt.month_name()

plt.figure(figsize=(8, 8))
df["Month"].value_counts().reindex([
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]).dropna().plot.pie(autopct="%1.1f%%")
plt.title("Month-wise Patient Admissions")
plt.ylabel("")
plt.show()
```

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[0, 17, 35, 50, 65, 100],
    labels=["0-17", "18-35", "36-50", "51-65", "66+"]
)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Age Group",
    hue="Age Group",
    palette="Blues",
    legend=False
)

plt.title("Patient Count by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Patient Count")
plt.tight_layout()
plt.show()
```
